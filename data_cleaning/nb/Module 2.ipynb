Calculate which laptops offer the best specs for the price.

What is the ValueScore?
It's a custom metric that answers: "How much laptop am I getting for my money?"

How it works:

1. Numerator (top of the fraction) - Combines all the good specs:
   - `RAM` - More RAM is better (in GB, like 8 or 16)
   - `HD/100` - Hard drive size divided by 100 (so 500GB becomes 5)
     - Why divide? So HD doesn't dominate the score since it's in the hundreds
   - `ScreenSize*10` - Screen size multiplied by 10 (so 15.6 inches becomes 156)
     - Why multiply? So screen size has more weight in the final score

2. Denominator (bottom) - The price:
   - Dividing by price means: higher specs at lower prices = better value

```python
# Calculate which laptops offer the best specs for the price
new_df['ValueScore'] = (new_df['RAM'] + new_df['HD']/100 + new_df['ScreenSize']*10) / new_df['Price']
best_value = new_df.nlargest(5, 'ValueScore')[['Brand', 'Price', 'RAM', 'HD', 'ScreenSize', 'ValueScore']]

print("\n")
print("Top 5 Best Value Laptops (Most specs per dollar)")
print(best_value.round(3))
```
