```python
import pandas as pd

data = pd.read_excel('./Inpatient Data (3).xlsx')
df = pd.DataFrame(data)
df.head(3)
```

## Viz
### Boxplot

```python
## Boxplot: Compare the distribution of a continuous variable across categories, compare hospital bill across carelevel
## Use X and Y
## X categorical Y continous

import matplotlib as plt
import seaborn as snb

ax = snb.boxplot(
    data=df,
    x='CareLevel',
    y='HospitalBill',
    color='lightgreen'
)

ax.set_title('Hospital Bill by Care Level')
ax.set_xlabel('Care Level')
ax.set_ylabel('Hopistal Bill')
```

```python
ax_2 = snb.boxplot(
    data=df,
    x='CareLevel',
    y='age'
)

ax_2.set_title('Age per Care Level')
ax_2.set_xlabel('Care Level')
ax_2.set_ylabel('Age')
```

```python
## Bar plot: Show the count of patients in each category of a categorical variable, bar of patient per discharge status or admission source
## Count, rename and call that in barplot

counts = df['Discarge_status'].value_counts().reset_index() # this creates the counts column, and reset index creates the df for columns
counts.columns = ['Discarge_status','patient_count']        # this renames

ax = snb.barplot(
    data=counts,
    x='Discarge_status',
    y='patient_count'
)

ax.set_title('Patient count per Discharge Status')
```

```python
counts_as = df['AdmissionSource'].value_counts().reset_index()
counts_as.columns = ['admission_source','patient_count'] # rename

ax = snb.barplot(
    data = counts_as,
    x = 'admission_source',
    y = 'patient_count'
)

ax.set_title('Patient count per Admission Source')
```

```python
## Histogram
## Mostly use data and one variable and set binwidth

# Hospital Bill histogram
ax = snb.histplot(
    data=df,
    x='HospitalBill',
    binwidth=1000,
)

ax.set_title('Histogram for Hospital Bill')
```

```python
# Histogram Histogram

ax = snb.histplot(
    data=df,
    x='age',
    binwidth=5,
)

ax.set_title('Histogram for Age')
```

```python
# Comorbidity Count Histogram

ax = snb.histplot(
    data=df,
    x='ComorbidityCount',
    discrete=True
)

ax.set_title('Histogram for Comorbidity')
```

```python
## Scaterplot: Show relationship between two numerical variables, possibly colored by third ( Ex: Age vs hospital bill colored by carelevel)
## Choose variables and third variable with hue

ax = snb.scatterplot(
    data=df,
    x='age',
    y='HospitalBill',
    hue='CareLevel',
    alpha=0.6,
    s=45,
    palette='Set2'
)

ax.set_title('Hospital Bill vs Age by Care Level')
ax.set_xlabel('Age')
ax.set_ylabel('Hospital Bill')
```
